## Setup

Run the cell below to load the dataset and initialize the summary DataFrame.

````python
import pandas as pd

# Load the dataset
df = pd.read_csv("capes_dataset_cleaned.csv")

# Initialize empty summary DataFrame
summary = pd.DataFrame(columns=[
    'Class',
    'Num Instances',
    'Avg Enrolled',
    'Avg Study Hours',
    'Avg GPA Expected',
    'Std GPA Expected',
    'Avg Recommend Class',
    'Avg CAPES Response Rate'
])

# Display the first few rows and columns
print("Columns:", df.columns.tolist())
df.head()
````

## Example: CSE 11

Below is a complete example for CSE 11. The code:
1. Filters the dataset for CSE 11 course instances
2. Removes instances with fewer than 20 students enrolled
3. Removes instances with less than 20% CAPES response rate
4. Calculates 7 summary statistics
5. Appends the results to the summary DataFrame

**Study this example carefully** — you will replicate this for the other 4 courses.

````python
# === CSE 11 ===

# Filter for only CSE 11 course instances
cse11_data = df[df['Course'] == "CSE 11"]

# Filter out instances with fewer than 20 students enrolled
cse11_data = cse11_data[cse11_data['Total Enrolled'] > 20]

# Filter out instances with less than 20% CAPES response rate
cse11_data = cse11_data[cse11_data['Total CAPEs Given'] > cse11_data['Total Enrolled'] * 0.2]

# Calculate summary statistics
num_instances_11 = len(cse11_data)
avg_enrolled_11 = cse11_data['Total Enrolled'].mean()
avg_hours_11 = cse11_data['Study Hours per Week'].mean()
avg_gpa_11 = cse11_data['Average GPA Expected'].mean()
std_gpa_11 = cse11_data['Average GPA Expected'].std()
avg_recommend_11 = cse11_data['Percentage Recommended Class'].mean()
avg_response_rate_11 = (cse11_data['Total CAPEs Given'] / cse11_data['Total Enrolled']).mean()

# Append to summary
summary.loc[len(summary)] = {
    'Class': 'CSE 11',
    'Num Instances': num_instances_11,
    'Avg Enrolled': avg_enrolled_11,
    'Avg Study Hours': avg_hours_11,
    'Avg GPA Expected': avg_gpa_11,
    'Std GPA Expected': std_gpa_11,
    'Avg Recommend Class': avg_recommend_11,
    'Avg CAPES Response Rate': avg_response_rate_11
}

print(f"CSE 11: {num_instances_11} instances after filtering")
summary
````
