````python
import pandas as pd
# Load the dataset
df = pd.read_csv("capes_dataset_cleaned.csv")

# Initialize empty summary DataFrame
summary = pd.DataFrame(columns=[
    'Class',
    'Average Enrolled',
    'Average study hours per week',
    'Average grade received',
    'Average CAPES response rate'
])

# Display the first few rows
df.head()
````

Filtering the cse11_dataset to only contain course instances of CSE 11.

````python

# Filter for only CSE 11 course instances
cse11_data = df[df['Course'] == "CSE 11"]

cse11_data.head()
````

Filtering out all instances that have less than a 20% CAPES response rate. Filter out all instances that have less than 20 students enrolled.

````python

cse11_data = cse11_data[cse11_data['Total Enrolled'] > 20]

# Question to ask, should we just clean total capes given to be %?
# instead of having the user calculate the total capes %?
cse11_data = cse11_data[cse11_data['Total CAPEs Given'] > cse11_data['Total Enrolled'] * 0.2]


cse11_data.head()
````

Calculating the average amount of hours worked, the average GPA received, the average number of students enrolled per each instance, and lastly the average CAPES response rate. Save each of these as separate variables.

````python
# Calculate average hours worked per week
avg_hours_11 = cse11_data['Study Hours per Week'].mean()

# Average GPA
avg_grade_11 = cse11_data['Average GPA Expected'].mean()

# Average number of students enrolled
avg_enrolled_11 = cse11_data['Total Enrolled'].mean()

# Average CAPES response rate
avg_response_rate_11 = (cse11_data['Total CAPEs Given'] / cse11_data['Total Enrolled']).mean()

print(f"Avg Hours Worked (CSE 11): {avg_hours_11:.2f}")
print(f"Avg Grade Expected (CSE 11): {avg_grade_11:.2f}")
print(f"Avg Enrolled (CSE 11): {avg_enrolled_11:.2f}")
print(f"Avg CAPES Response Rate (CSE 11): {avg_response_rate_11:.2%}")
````

Appending the relevant information to the summary data set.

````python
# Create a summary DataFrame with one row for CSE 11
summary.loc[len(summary)] = {
    'Class': 'CSE 11',
    'Average Enrolled': avg_enrolled_11,
    'Average study hours per week': avg_hours_11,
    'Average grade received': avg_grade_11,
    'Average CAPES response rate': avg_response_rate_11
}

summary
````
