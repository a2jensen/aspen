```python
import matplotlib.pyplot as plt
import pandas as pd

# Plot figures directly in the notebook
%matplotlib inline
```

PART 1:
You are given a file gdp_dataset.csv which contains the list of countries, along with their GDP per capita for each year. The dataset is already cleaned, containing only the necessary countries and years, with columns being years and rows being countries.
Based on this file, create one line graph for each country (there should be 12 total).

```python
df = pd.read_csv("gdp_dataset.csv")
df
```

```python
argentina_data = df[df['country'] == 'Argentina']
years = df.columns[1:]
gdp_argentina = argentina_data.iloc[0,1:]
fig = plt.figure()
ax = plt.axes()
ax.plot(years, gdp_argentina)
plt.title('Argentina GDP Per Capita Over Time')
ax.set_xlabel('Year')
ax.set_ylabel('GDP per capita (current US$)')
#ax.legend(loc='upper right')
```

```python

brazil_data = df[df['country'] == 'Brazil']
years = df.columns[1:]
gdp_brazil = brazil_data.iloc[0,1:]
fig = plt.figure()
ax = plt.axes()
ax.plot(years, gdp_brazil)
plt.title('Brazil GDP Per Capita Over Time')
ax.set_xlabel('Year')
ax.set_ylabel('GDP per capita (current US$)')
```

```python
countries = df['country'].tolist()
for country in countries:
    country_data = df[df['country'] == country]
    gdp_values = country_data.iloc[0,1:]
    fig = plt.figure()
    ax = plt.axes()
    ax.plot(years, gdp_values)
    plt.title(f'{country} GDP Per Capita Over Time')
    ax.set_xlabel('Year')
    ax.set_ylabel('GDP per capita (current US$)')
```

PART 2:
Now, create one line graph displaying the GDP per capita from 2012 to 2022, with a separate line for each country.

```python
#countries = df['country'].tolist()

fig = plt.figure()
ax = plt.axes()
plt.title('GDP Per Capita Over Time for South American Countries')
ax.set_xlabel('Year')
ax.set_ylabel('GDP per capita (current US$)')

for country in countries:
    country_data = df[df['country'] == country]
    gdp_values = country_data.iloc[0,1:]
    ax.plot(years, gdp_values, label=country)

plt.legend()
```
