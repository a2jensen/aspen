You are working on a project aiming to get insights on class information for a few of the lower division computer science course offerings here at UCSD.

You will be working with the CAPES dataset. You will compute a specific statistic for each of the following classes:

- **CSE 11** → total students enrolled (across all course instances)
- **CSE 12** → total CAPEs given (across all course instances)
- **CSE 30** → average study hours per week

For context, when we say a course instance we mean a course offering for a certain quarter. Examples include ["CSE 12 SP23", "CSE 12 WI23", "CSE 12 FA18", etc.]

````python
import pandas as pd

# Load the dataset
df = pd.read_csv("capes_dataset_cleaned.csv")

# Display the first few rows
df.head()
````

Filtering the cse11_dataset to only contain course instances of CSE 11.

````python

# Filter for only CSE 11 course instances
cse11_data = df[df['Course'] == "CSE 11"]

cse11_data.head()
````

Filtering out all instances that have less than a 20% CAPES response rate. Filter out all instances that have less than 20 students enrolled.

````python

cse11_data = cse11_data[cse11_data['Total Enrolled'] > 20]

# Question to ask, should we just clean total capes given to be %?
# instead of having the user calculate the total capes %?
cse11_data = cse11_data[cse11_data['Total CAPEs Given'] > cse11_data['Total Enrolled'] * 0.2]


cse11_data.head()
````

Calculate the total number of students enrolled across all CSE 11 course instances. Save this as a variable.

````python
# Total students enrolled across all CSE 11 course instances
total_enrolled_11 = cse11_data['Total Enrolled'].sum()

print(f"CSE 11 - Total Enrolled: {total_enrolled_11}")
````

Display the CSE 11 summary.

````python
pd.DataFrame({
    'Class': ['CSE 11'],
    'Total Enrolled': [total_enrolled_11]
})
````

Now apply the same filtering steps to the remaining classes below, then compute the specific statistic for each one.

- **CSE 12**: total CAPEs given
- **CSE 30**: average study hours per week

````python
# CSE 12

# Filter for only CSE 12 course instances
cse12_data = df[df['Course'] == "CSE 12"]

cse12_data = cse12_data[cse12_data['Total Enrolled'] > 20]
cse12_data = cse12_data[cse12_data['Total CAPEs Given'] > cse12_data['Total Enrolled'] * 0.2]

# Total CAPEs given across all CSE 12 course instances
total_capes_given_12 = cse12_data['Total CAPEs Given'].sum()

print(f"CSE 12 - Total CAPEs Given: {total_capes_given_12}")

pd.DataFrame({
    'Class': ['CSE 12'],
    'Total CAPEs Given': [total_capes_given_12]
})
````

````python
# CSE 30

# Filter for only CSE 30 course instances
cse30_data = df[df['Course'] == "CSE 30"]

cse30_data = cse30_data[cse30_data['Total Enrolled'] > 20]
cse30_data = cse30_data[cse30_data['Total CAPEs Given'] > cse30_data['Total Enrolled'] * 0.2]

# Average study hours per week across all CSE 30 course instances
avg_study_hours_30 = cse30_data['Study Hours per Week'].mean()

print(f"CSE 30 - Avg Study Hours/Week: {avg_study_hours_30:.2f}")

pd.DataFrame({
    'Class': ['CSE 30'],
    'Avg Study Hours/Week': [avg_study_hours_30]
})
````
