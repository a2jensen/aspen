```python
# Make a copy to work on
import pandas as pd

# Load the dataset
df = pd.read_csv("capes_dataset.csv")

df_clean = df.copy()

# Clean the Course column + remove Evaluation column
df_clean['Course'] = df_clean['Course'].str.split('-').str[0].str.strip()
df_clean = df_clean.drop(['Evalulation URL'], axis=1)

# Clean grade columns to include only GPA
df_clean['Average Grade Expected'] = df_clean['Average Grade Expected'].str.extract(r'\(([\d.]+)\)').astype(float)
df_clean['Average Grade Received'] = df_clean['Average Grade Received'].str.extract(r'\(([\d.]+)\)').astype(float)
df_clean = df_clean.rename(columns={"Average Grade Expected": "Average GPA Expected", 
                   "Average Grade Received": "Average GPA Received"})
df_clean = df_clean.rename(columns={"Total Enrolled in Course" : "Total Enrolled"})

# Display result
df_clean.head()


# Save cleaned version
df_clean.to_csv("capes_dataset_cleaned.csv", index=False)
```
